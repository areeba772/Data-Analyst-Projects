```python
# Dataset load karein
df = pd.read_csv('Data/Cleaned_Food_Delivery_Data.csv')

# Dashboard layout (2 rows, 2 columns) banana
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 10))
fig.suptitle('Food Delivery Operations Analytics Dashboard', fontsize=16, fontweight='bold')

# 1. Delivery Time Distribution (Top-Left)
sns.histplot(df['Delivery_Time_Min'], kde=True, color='orange', ax=axes[0, 0])
axes[0, 0].set_title('Delivery Time Distribution')
axes[0, 0].set_xlabel('Delivery Time (Min)')
axes[0, 0].set_ylabel('Order Count')

# 2. Total Revenue by City (Top-Right)
sns.barplot(x='City', y='Revenue_USD', data=df, estimator=sum, palette='viridis', ax=axes[0, 1])
axes[0, 1].set_title('Total Revenue by City')
axes[0, 1].set_xlabel('City')
axes[0, 1].set_ylabel('Total Revenue (USD)')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Customer Churn Risk vs Order Value (Bottom-Left)
sns.scatterplot(x='Order_Value_USD', y='Churn_Risk', hue='Customer_Type', data=df, alpha=0.6, ax=axes[1, 0])
axes[1, 0].set_title('Churn Risk vs Order Value')
axes[1, 0].set_xlabel('Order Value (USD)')
axes[1, 0].set_ylabel('Churn Risk Score')

# 4. Delivery Distance vs Delivery Time (Bottom-Right)
sns.scatterplot(x='Delivery_Distance_KM', y='Delivery_Time_Min', color='green', alpha=0.5, ax=axes[1, 1])
axes[1, 1].set_title('Delivery Distance vs Delivery Time')
axes[1, 1].set_xlabel('Distance (KM)')
axes[1, 1].set_ylabel('Delivery Time (Min)')

# Layout adjust karna aur save karna
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('Data/delivery_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 3. Graph: Delivery Time Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df['Delivery_Time_Min'], kde=True, color='orange')
plt.title('Food Delivery Time Distribution')
plt.xlabel('Delivery Time (Minutes)')
plt.ylabel('Order Count')
plt.show()
```

```python
# Graph 1: Delivery Time Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df['Delivery_Time_Min'], kde=True, color='orange')
plt.title('Food Delivery Time Distribution')
plt.xlabel('Delivery Time (Minutes)')
plt.ylabel('Order Count')

# ---> Yahan graph save hoga (plt.show() se pehle) <---
plt.savefig('Data/delivery_time_distribution.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Graph 2: Total Revenue by City
plt.figure(figsize=(10, 5))
sns.barplot(x='City', y='Revenue_USD', data=df, estimator=sum, palette='viridis')
plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Total Revenue (USD)')
plt.xticks(rotation=45)

# Save Graph 2
plt.savefig('Data/revenue_by_city.png', dpi=300, bbox_inches='tight')
plt.show()

# Graph 3: Customer Churn Risk vs Order Value
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Order_Value_USD', y='Churn_Risk', hue='Customer_Type', data=df, alpha=0.7)
plt.title('Customer Churn Risk vs Order Value')
plt.xlabel('Order Value (USD)')
plt.ylabel('Churn Risk Score')

# Save Graph 3
plt.savefig('Data/churn_risk_vs_order_value.png', dpi=300, bbox_inches='tight')
plt.show()
```
